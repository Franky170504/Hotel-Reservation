```python
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv(r"data\raw\preprocessed_data.csv")
df
```

```python
from sklearn.ensemble import RandomForestClassifier
```

```python
X = df.drop(columns="booking_status")
y = df["booking_status"]
```

```python
model = RandomForestClassifier(random_state=42)
model.fit(X,y)
```

```python
feature_importance = model.feature_importances_
```

```python
feature_importance_df = pd.DataFrame({
    'feature' : X.columns,
    'importance' :feature_importance
})
```

```python
top_feature_importance_df=feature_importance_df.sort_values(by="importance", ascending=False)
```

```python
top_feature_importance_df
```

```python
top_10_features = top_feature_importance_df["feature"].head(10).values

top_10_df = df[top_10_features.tolist() + ["booking_status"]]
```

```python
top_10_df
```

```python
if os.path.exists(r"data\raw\top_features.csv"):
    print("File already exists")
else:
    top_10_df.to_csv(r"data\raw\top_features.csv", index= False)
```
